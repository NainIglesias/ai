# Operaciones con DataFrames

````
import pandas as pd
df = pd.read_csv('data/grades.csv', index_col=0)
df
````

## Gestión de columnas

````
df.insert(0, 'ID', ['11222333A', '22333222B', '33222333C', '44333222D'])  # inserta una columna ID en la posición 0
df.insert(1, 'Age', [23, 23, 36, 36])  # inserta una columna Age en la posición 1
df
````

````
# O simplemente podemos añadir una columna al final
df['Other'] = [1, 1, 1, 1]  # si la columna no existe, la crea; si existe, la sobrescribe
df
````

````
# Crea una columna nueva con las notas de PIA normalizadas
df['PIA_norm'] = (df['PIA'] - df['PIA'].min()) / (df['PIA'].max() - df['PIA'].min())
df
````

````
df.drop('Other', axis=1)  # devuelve un DataFrame nuevo sin la columna 'Other', pero no modifica el DataFrame original
# df  # si mostramos el DataFrame original, vemos que la columna 'Other' sigue ahí
````

```drop``` tiene el parámetro ```axis```, que por defecto vale 0, lo que indica que se borrarán filas. Si quieres borrar columnas, tienes que especificar ```axis=1```, o su equivalente ```axis='columns'```.

````
df.drop('Other', axis=1, inplace=True)  # borra la columna 'Other' del DataFrame original
df
````

> [!IMPORTANT]
> Con `inplace=True`, el método modifica el propio DataFrame y devuelve `None`, en lugar de devolver un DataFrame nuevo. Como devuelve `None`, no se puede encadenar con otros métodos. Los desarrolladores de pandas lo desaconsejan: han aceptado una propuesta, [PDEP-8](https://pandas.pydata.org/pdeps/0008-inplace-methods-in-pandas.html), para eliminarlo de la mayoría de los métodos, como `drop`, `rename` y `set_index`, porque asignar el resultado hace lo mismo. Asigna el resultado en su lugar: `df = df.drop(columns='Other')`.

También puedes usar los parámetros ```columns``` e ```index``` para borrar columnas o filas, respectivamente.

````
df.drop(columns='Age')  # devuelve otro DataFrame sin 'Age'
````

````
df.drop(columns=['PIA_norm', 'Age'])  # devuelve otro DataFrame sin 'PIA_norm' ni 'Age'
````

### Usar `assign` para añadir columnas

El método `assign` ofrece una forma más limpia de añadir columnas calculadas, sobre todo al encadenar métodos. A diferencia de la asignación directa (`df['col'] = ...`), `assign` devuelve un DataFrame nuevo sin modificar el original.

````
# assign devuelve un DataFrame nuevo con la columna nueva
df.assign(PIA_doubled=df['PIA'] * 2)
````

````
# Usar lambda permite hacer referencia a columnas creadas en la misma llamada a assign
df.assign(
    average=lambda x: (x['PIA'] + x['SAA'] + x['MIA']) / 3,
    passed=lambda x: x['average'] >= 5  # Usa la columna que se acaba de crear arriba
)
````

> **Consejo:** `assign` es especialmente potente combinado con `groupby` y `transform` para añadir columnas calculadas a partir de estadísticos de grupo. Consulta el notebook **[Transformaciones de DataFrames](pandas_transformations.ipynb)** para ver ejemplos avanzados.

## Gestión de filas

````
# df.at['Alan Turing'] # Daría error; at solo sirve para acceder a una celda concreta
df.loc['Alan Turing'] # Obtiene una fila completa como Series
````

````
import numpy as np
df.loc['Arthur Samuel'] = ['55444333E', 25, 8, 8.0, 7.0, 9.0, 6.0, np.nan] # Añade un alumno nuevo
# Como la columna PIA_norm necesita antes la columna PIA, no podemos añadir el valor de PIA_norm
df
````

El valor de la columna "PIA_norm" tiene que calcularse procesando toda la columna "PIA" y asignando después el resultado a la columna "PIA_norm". Por eso he optado por asignar la **constante** ```np.nan``` **(Not A Number)** al añadir una fila nueva: para indicar que de momento no tenemos esa información (tendremos que recalcular los valores normalizados).

````
df.drop('Arthur Samuel') # Las filas se pueden borrar como las columnas (en este caso no uso inplace=True porque no quiero modificar el DataFrame original)
#df
````

## Acceder a los elementos

````
print(df.at['Alan Turing','PIA']) # Accede a una celda concreta de forma más eficiente que loc
print(df.loc['Alan Turing','PIA'])
print(df.loc['Alan Turing'].at['PIA'])  # loc devuelve una Series con los datos de la fila, así que podemos usar at para acceder a un elemento concreto de esa Series
print(df['PIA']['Alan Turing'])
print(df['PIA'].at['Alan Turing'])
print(df.iat[0,2])  # Acceso por posición entera: primera fila, tercera columna
print(df.iloc[0,2])
````

`.at` lee o escribe un valor por sus etiquetas, y es la forma más rápida de hacerlo; `.loc` también selecciona por etiquetas, pero además puede devolver una fila, una columna o un slice. `.iat` e `.iloc` hacen lo mismo por posición.

La tercera y la cuarta línea son *indexación encadenada*: una primera selección devuelve una Series, y una segunda coge el valor de ella. Leer así funciona, pero es más lento y menos claro que un único `.at` o `.loc`. Escribir así no funciona en absoluto, como muestran las celdas siguientes.

````
df.at['Alan Turing', 'PIA'] = 10 # Modifica un valor usando el índice de la fila y el nombre de la columna
df.loc['Alan Turing', 'PIA'] = 10 # Equivale a lo anterior
df
````

````
df.iloc[1, 2] = 0 # Modifica una nota usando posiciones
df
````

Desde pandas 3.0, toda selección se comporta como una copia (*copy-on-write*), así que un valor asignado mediante indexación encadenada cambia esa copia temporal y nunca `df`. pandas lo avisa con un `ChainedAssignmentError`, y el valor se queda como estaba. Para asignar, selecciona con un único `.loc` o `.at`, como arriba:

````
df['PIA']['Alan Turing'] = 0  # asignación encadenada: pandas avisa, y df no cambia
df.loc['Alan Turing', 'PIA']
````

Cuando las etiquetas son enteros, `.loc` e `.iloc` dan resultados distintos: `.loc[1, 1]` es la fila y la columna con *etiqueta* 1, e `.iloc[1, 1]` la segunda fila y la segunda columna.

````
dict_num_keys1 = {1: 'one', 2: 'two', 3: 'three'}
dict_num_keys2 = {1: 'I', 2: 'II', 3: 'III'}
dict_num_keys3 = {1: 'first', 2: 'second', 3: 'third'}

df_ints_as_keys = pd.DataFrame({1: dict_num_keys1, 2: dict_num_keys2, 3: dict_num_keys3})
print(df_ints_as_keys.iloc[1, 1])
print(df_ints_as_keys.loc[1, 1])
````

````
df[['ID']]  # devuelve un DataFrame con la columna ID
````

````
df['ID']  # devuelve una Series con la columna ID
````

````
df.ID  # equivale a lo anterior
````

## Slicing

````
df.loc[:, 'PIA':'MIA'] # Devuelve un DataFrame con todas las filas y las columnas PIA, SAA y MIA
````

````
df.iloc[:3, 0:3] # Devuelve un DataFrame con las filas anteriores a la fila 3, y las columnas 0, 1 y 2
````

````
df.loc[:'Marvin Minsky'] # Devuelve un DataFrame con las filas hasta Marvin Minsky, incluida
````

## Crear un DataFrame nuevo aplicando filtros

````
# Alumnos que aprobaron PIA y SAA
df[(df['PIA'] >= 5) & (df['SAA'] >= 5)]
````

````
df[(df['PIA'] >= 5) | (df['SAA'] >= 5)] [['PIA','SAA','MIA']] # Alumnos que aprobaron PIA o SAA, con sus notas en esos módulos y en MIA
````

````
passed_pia_and_saa = df[(df['PIA'] >= 5) & (df['SAA'] >= 5)]  # alumnos que aprobaron PIA y SAA
passed_pia_and_saa = passed_pia_and_saa.set_index('ID')  # identificados por su ID
passed_pia_and_saa[['PIA', 'SAA', 'MIA']]  # con sus notas en esos módulos y en MIA
````

````
df[(df['PIA'] >= 5) & (df['SAA'] >= 5)].set_index('ID')[['PIA', 'SAA', 'MIA']]  # lo mismo, en una sola línea
````

## Ordenar

````
df.sort_values('PIA', ascending=False) # Ordena el DataFrame por la columna PIA en orden descendente
````

````
df.sort_values(by=['PIA', 'SAA', 'MIA'], ascending=False) # Ordena el DataFrame por las columnas PIA, SAA y MIA (en ese orden de prioridad)
````

## Agregaciones básicas

Pandas ofrece métodos sencillos para calcular estadísticos sobre columnas o filas.

````
df['PIA'].mean() # Media de la columna PIA
````

````
df.sum() # Suma de cada columna
````

````
df.loc[:, ['PIA', 'SAA', 'MIA', 'BDA', 'SBD']].mean(axis='columns')  # nota media de cada alumno en los cinco módulos
# axis='columns' (o axis=1) calcula la media a lo largo de las columnas, una vez por cada fila
````

````
df['SBD'].value_counts() # Devuelve el número de veces que aparece cada valor en la columna SBD
# cuatro alumnos sacaron un 9 en SBD y uno sacó un 10
````

````
df.isnull() # Devuelve un DataFrame con True en las celdas que contienen NaN y False en las que no
df.isna()  # Equivale a lo anterior («not available», no disponible)
````

````
df.isnull().sum() # Devuelve el número de valores nulos de cada columna
````

Para las **operaciones de agrupación** (`groupby`, `agg`, `transform` y `apply` por grupos), las **transformaciones elemento a elemento** (`map`, `apply`, `transform`) y el **remodelado de datos** (`melt`, `pivot`, `pivot_table`), consulta el notebook **[Transformaciones y remodelado de DataFrames](pandas_transformations.ipynb)**.
