# NumPy

## Introducción

**NumPy** es una biblioteca de Python que proporciona funciones para trabajar con arrays de 1 (vectores), 2 (matrices) y n dimensiones.

NumPy funciona de forma parecida a las listas de Python, pero con la diferencia de que sus arrays son homogéneos (todos los elementos son del mismo tipo) y tienen un tamaño fijo (sus elementos pueden cambiar, pero el array no puede crecer ni encogerse).

Los arrays de NumPy son más eficientes que las listas de Python, ya que están implementados en C y se ejecutan más rápido. Usan mucha menos memoria y permiten especificar el tipo de dato que contiene el array.

Es una de las bibliotecas más usadas en el campo de la ciencia de datos y el aprendizaje automático. Muchas otras bibliotecas la usan como base (es por tanto una **dependencia** para ellas): **pandas, Matplotlib, Seaborn, scikit-learn, TensorFlow, PyTorch, Keras**...

## Tipos de datos de los elementos de un array

Los objetos array de NumPy son siempre de tipo ndarray ([*N-dimensional array*](https://numpy.org/doc/stable/reference/arrays.ndarray.html), array N-dimensional). El tipo de sus elementos lo especifica su dtype ([*data type*](https://numpy.org/doc/stable/reference/arrays.dtypes.html#arrays-dtypes), tipo de dato).
Los ndarrays son siempre homogéneos, es decir, todos sus elementos son del mismo tipo.

````
# Si no especificamos el tipo de los elementos, NumPy lo deduce a partir del argumento
nested_list = [[1, 2, 3], [4, 5, 6]]
a = np.array(nested_list)
print(a)
print(a.dtype)

# Podemos crear un ndarray de NumPy con elementos de un tipo concreto
b = np.array(nested_list, dtype=np.float64) # float de 64 bits
# Aunque pasamos una lista con ints, con el parámetro dtype podemos indicar que queremos que los elementos sean floats, así que los convierte. Lo vemos al mostrarlo con print por el punto decimal.
print(b)
print(b.dtype)
````

````
# Supongamos que usamos un array como el siguiente para guardar las notas de 2 exámenes de una lista de clase:
students_and_grades = np.array([['Antonio','Bea','Carlos','Diana'],
                            [65,78,90,81],
                            [71,82,79,92]]) # ¿Ves algo que se podría mejorar en esta estructura de datos?

# Los elementos de un array de NumPy deben ser del mismo tipo. Si pasamos una lista con elementos de distintos tipos, NumPy los convierte todos al mismo tipo. En este caso, como hay strings, los convierte todos a strings.
print(students_and_grades)
print(students_and_grades.dtype) # dtype='<U11' significa que son strings Unicode de hasta 11 caracteres
````

Aparte de que el ejemplo anterior sirve para ilustrar los tipos en NumPy, esa estructura se podría mejorar. No solo no encaja con la filosofía de NumPy (ejecución eficiente de operaciones matemáticas); además, no cambiaría mucho respecto a dejarla como lista de listas:

```python
students_and_grades = [['Antonio', 'Bea', 'Carlos', 'Diana'],
                     [65, 78, 90, 81],
                     [71, 82, 79, 92]]
```

pero en lugar de enlazar varias listas por su índice (lo que puede ser fuente de errores al borrar u ordenar elementos de forma sincronizada en todas ellas), lo recomendable sería encapsular primero cada alumno con sus notas (en una tupla o un objeto) y mantener una lista de ellos:

```python
students_and_grades = [('Antonio', [65, 71]), ('Bea', [78, 82]), ('Carlos', [90, 79]), ('Diana', [81, 92])]
```

Esto sería discutible si necesitamos hacer un procesamiento costoso de un gran número de notas de alumnos, pero en ese caso lo recomendable sería usar una estructura de datos más compleja, como un **DataFrame de Pandas**, que nos permitiría hacer las operaciones de forma eficiente.

### NaN

El tipo de dato **float** en NumPy admite el valor especial **NaN** (*Not a Number*, «no es un número»), que se usa para representar valores numéricos que no son números reales manteniendo el tipo float, para poder hacer operaciones con ellos. Por ejemplo, el resultado de dividir 0 entre 0 es NaN.

Si usáramos el tipo None para representar valores numéricos que no son números reales, el tipo de dato resultante sería object, y no podríamos hacer operaciones matemáticas con él.

````
print(type(np.nan))
print(type(None))
````

````
np.array([[1, 2, 3], [4, np.nan, 6]])
````

````
A = np.array([[11, 12, 13, 14], [21, 22, 23, 24], [31, 32, 33, 34]])
A
````

La celda anterior define con NumPy una matriz de 3x4 (3 filas y 4 columnas) como la siguiente:

$$
\begin{pmatrix}
    11 & 12 & 13 & 14\\
    21 & 22 & 23 & 24\\
    31 & 32 & 33 & 34
\end{pmatrix}
$$

Para acceder a sus valores en Python se usa la sintaxis ``` A[fila, columna]```, recordando que los índices empiezan en 0, así que al primer elemento de la matriz se accede con ```A[0,0]``` y al último con ```A[2,3]```.

$$
\begin{array}{c c}
    & \color{red} \begin{matrix}
    col\ 0\  &\ \ col\ 1\  &\  \ col\ 2\  &\  \ col\ 3 \end{matrix}
    \\
    \color{red}
    \begin{matrix} fila\ 0 \\ fila\ 1 \\ fila\ 2 \end{matrix}
    &
    \begin{pmatrix}
        A[0,0] & A[0,1] & A[0,2] & A[0,3]\\
        A[1,0] & A[1,1] & A[1,2] & A[1,3]\\
        A[2,0] & A[2,1] & A[2,2] & A[2,3]
    \end{pmatrix}
\end{array}
$$

````
print(A[0,0]) # Elemento de la primera fila (fila 0) y la primera columna (columna 0)
print(A[1,2]) # Elemento de la segunda fila (fila 1) y la tercera columna (columna 2)
print(A[1]) # Segunda fila de A (índice 1)
````

## Funciones básicas para crear arrays

````
print(np.zeros(4)) # Crea un vector de NumPy con cuatro elementos a 0
print(np.ones(2)) # Crea un ndarray de NumPy con dos elementos a 1
````

````
# El parámetro shape nos permite especificar la forma de la matriz; recibe una tupla con sus dimensiones
np.zeros(shape=(2,3)) # Crea una matriz de ceros de 2x3. Dimensiones (tupla 2,3) con ceros
````

````
np.ones((2,2,2), dtype=int) # Crea un array tridimensional de 2x2x2 con unos de tipo int
````

````
# np.zeros y np.ones usan por defecto el tipo float64, pero podemos especificar el tipo con el parámetro dtype
print(np.zeros(4))
print(np.zeros(4, dtype=np.int64))

print(np.arange(254, 259, dtype=np.uint8)) # En este caso el tipo es uint8 (int de 8 bits sin signo), así que el valor máximo que puede tomar es 255 y se desborda
````

````
# np.empty no inicializa sus elementos, así que toman el valor que hubiera en esa posición de memoria
np.empty(2)
````

````
print(np.full((3,3), True)) # Crea una matriz de 3x3 con todos los elementos a True
print(np.full(shape=(3,2), fill_value=5)) # Crea una matriz de 3x2 con todos los elementos a 5 (int)

# np.full toma el tipo del argumento que le pasamos (igual que np.array con los elementos de la secuencia que recibe), así que si le pasamos un float, los elementos serán floats

print(np.full((3,2), 5.0)) # Crea una matriz de 3x2 con todos los elementos a 5.0
print(np.full((3,2), 5, dtype=np.float64)) # Crea una matriz de 3x2 con todos los elementos a 5.0
````

````
np.array([1, 2, 3]).reshape(-1,1) # Convierte un vector en una matriz de una sola columna
````

## Slicing (rebanado)

El *slicing* es una forma de obtener una parte de la matriz. Funciona de forma análoga a las listas de Python.

```
a[start:stop]  # elementos desde start hasta stop-1 (stop es el primer elemento no incluido)
a[start:]      # elementos desde start hasta el final del array
a[:stop]       # elementos desde el principio del array hasta stop-1
a[:]           # una copia del array completo
```

También hay un tercer valor, el incremento «step» (paso):
```
a[start:stop:step] # desde start hasta un valor menor que stop, con incremento «step»
```

Start y stop pueden ser números negativos. Eso significa contar desde el final del array:
```
a[-1]    # Último elemento del array
a[-2:]   # Últimos dos elementos del array
a[:-2]   # Todos los elementos del array excepto los dos últimos
```

Step también puede ser negativo:
```
a[::-1]    # todos los elementos del array pero al revés, empezando por el último (equivale a np.flip(a))
a[1::-1]   # los dos primeros elementos, al revés
a[:-3:-1]  # los dos últimos elementos, al revés
a[-3::-1]  # todo excepto los dos últimos elementos, al revés
```

[<img src="https://numpy.org/doc/stable/_images/np_matrix_indexing.png" width="800"/>](https://numpy.org/doc/stable/user/absolute_beginners.html#indexing-and-slicing)

````
print(f"{A}\n---")

print(f"Row 1: {A[1,:]}") # La fila con índice 1 y todas las columnas. Es decir: la 2.ª fila
print(f"Row 1: {A[1]}") # lo mismo que la anterior, simplificado
````

````
A[:,1] # Los elementos de todas las filas (:) y la columna 1. Es decir: los de la 2.ª columna
# Se devuelve un vector con esos valores (no una matriz con una sola columna)
````

````
# Igual que el anterior, pero devuelve una matriz de una sola columna
A[:,1:2]
# print(f"Column 1: {A[:,1].reshape(-1,1)}") # Equivalente
````

````
print(f"{A}\n---")

print(f"Row 1, columns 0 and 1: {A[1,:2]}") # Fila 1 y columnas 0 y 1 (todas hasta la 2)
````

````
print(f"{A}\n---")

print(A[1:3,1:3])  # Submatriz de A con las filas 1 y 2 y las columnas 1 y 2
print(A[1:,1:])  # Submatriz de A con las filas 1 y 2 y las columnas 1, 2 y 3
````

## Filtrado

````
arr = np.array([1, 2, 3, 4])
print(arr)
x = [True, False, True, False]
filtered_array = arr[x] # Filtra los elementos de arr que están en las posiciones True de x
print(filtered_array)
````

````

a = np.arange(11)**2
print(a)
print(a[a<10]) # Nos quedamos con los valores menores que 10
print(a[a==16]) # Nos quedamos con los valores iguales a 16
print(a[a!=16]) # Nos quedamos con los valores distintos de 16
print(a[(a>20) | (a<50)]) # Nos quedamos con los valores mayores que 20 o menores que 50
print(a[(a>20) & (a<50)]) # Nos quedamos con los valores mayores que 20 y menores que 50
````

````
# A partir de filtros podemos modificar ciertos elementos de un ndarray
a[a<10] = 0 # Pone a 0 los valores menores que 10
print(a)

b = np.arange(12).reshape(3,4)
print(b)
# Pone a 0 los valores impares de b
b[b%2==1] = 0
print(b)
````

## Operaciones matemáticas

[<img src="https://numpy.org/doc/stable/_images/np_aggregation.png" width="800"/>](https://numpy.org/doc/stable/user/absolute_beginners.html#more-useful-array-operations)

````
vector = np.array([9,10,1,2,2,3,3,3,4,5,6,7,8])

print(f"vector: {vector}")
print(f"Maximum: {vector.max()}")
print(f"Index of maximum: {vector.argmax()}")
print(f"Minimum: {vector.min()}")
print(f"Index of minimum: {vector.argmin()}")
print(f"Sum: {vector.sum()}")
print(f"Mean: {vector.mean()}")
print(f"Cumulative sum: {vector.cumsum()}")
````

````
print(f"Standard deviation: {vector.std():.3f}")
print(f"Variance: {vector.var():.3f}")
````

## Operaciones sobre matrices

### Ordenar

````
print(f"vector before sort:   {vector}")
print("--- Sorting procedurally ---")
# La función sort() recibe el vector como argumento y lo devuelve ordenado, pero no modifica el vector original
sorted_vec = np.sort(vector)
print(f"vector after sort: {vector}")
print(f"sorted:               {sorted_vec}")

print("--- Sorting object-oriented ---")
# El método sort() del objeto ndarray modifica el objeto desde el que se llama, y no devuelve nada
vector_copy = vector.copy() # copia del vector original
print(f"vector before sort:   {vector_copy}")
vector_copy.sort()
print(f"sorted vector:        {vector_copy}")
````

### Unique (valores únicos)

````
# numpy.unique() devuelve los valores únicos de un vector
unique_vals, positions = np.unique(vector, return_index=True) # return_index=True devuelve los índices de los valores únicos
print(f"vector:     {vector}")
print(f"unique:     {unique_vals}")
print(f"indices:    {positions}")
````

````
unique_vals, frequencies = np.unique(vector, return_counts=True) # return_counts=True devuelve las frecuencias de los valores únicos
print(f"vector:      {vector}")
print(f"unique:       {unique_vals}")
print(f"frequencies: {frequencies}")
# Si por ejemplo queremos un diccionario con cada valor y su frecuencia, podemos usar zip:
print(dict(zip(unique_vals, frequencies)))
# zip devuelve un iterador de tuplas que combina uno a uno los elementos de los iterables que le pasamos. En este caso combinamos los elementos de unique_vals y frequencies, que son dos arrays de NumPy, así que zip devuelve un iterador de tuplas de NumPy. Al pasarlo a dict, creamos un diccionario con las tuplas como pares clave-valor.
````
