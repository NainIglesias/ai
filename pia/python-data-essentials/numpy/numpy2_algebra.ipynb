# NumPy para operaciones algebraicas

```
import numpy as np
```

## Matriz identidad

```
print(np.identity(4), "\n --- ") # Crea una matriz identidad de 4x4
print(np.eye(4), "\n --- ") # Lo mismo que np.identity(4)

# np.identity llama a np.eye con k=0. np.eye permite crear matrices con unos en la diagonal desplazada
print(np.eye(4, k=1), "\n --- ") # Crea una matriz de 4x4 con unos en la diagonal desplazada 1 posición
```

## Matriz traspuesta

```
A = np.array([[1, 2, 3], [4, 5, 6]])
print(A)
print(A.T) # Traspuesta de A (intercambia filas y columnas)
```

```
# Ciertas operaciones de acceso con slices a una matriz se pueden simplificar usando la traspuesta:
print(A[:,1]) # Columna 1
print(A.T[1]) # Fila 1 de la traspuesta = columna 1
```

## Operaciones con matrices

### Operaciones de matrices con escalares

Se pueden aplicar operaciones aritméticas a una matriz y un escalar. El escalar se aplica a cada elemento de la matriz.

[<img src="https://numpy.org/doc/stable/_images/np_multiply_broadcasting.png" width="600">](https://numpy.org/doc/stable/user/absolute_beginners.html#broadcasting)

```
print(f"A=\n{A}")

print(f"A+3 =\n {A+3}") # Suma 3 a cada elemento de A
print(f"A-3 =\n {A-3}") # Resta 3 a cada elemento de A
print(f"A*2 =\n {A*2}") # Multiplica cada elemento de A por 2
print(f"A/2 =\n {A/2}") # Divide cada elemento de A entre 2
print(f"A**2 =\n {A**2}") # Eleva al cuadrado cada elemento de A
print(f"A**0.5 =\n {A**.5}") # Raíz cuadrada de cada elemento de A (eleva a 1/2)
print(f"√.5 =\n {np.sqrt(A)}") # Raíz cuadrada de cada elemento de A
```

### Operaciones elemento a elemento

Se pueden sumar, restar, multiplicar o dividir todos los elementos de dos matrices, cada uno con su elemento correspondiente.

[<img src="https://numpy.org/doc/stable/_images/np_sub_mult_divide.png" width="700"/>](https://numpy.org/doc/stable/user/absolute_beginners.html#basic-array-operations)

```
X = np.identity(2, dtype=int)
print(f"X =\n{X}")
Y = np.array([[1, 2], [3, 4]])
print(f"Y =\n{Y}")
print("---")
print(f"X + Y =\n{X + Y}") # Suma de matrices
print(f"X - Y =\n{X - Y}") # Resta de matrices
print(f"X * Y =\n{X * Y}") # Producto elemento a elemento
print(f"X/Y =\n{X/Y}") # División elemento a elemento
```

### Multiplicación de matrices (producto matricial)

La multiplicación de matrices consiste en multiplicar cada fila de la primera matriz por cada columna de la segunda, sumar los resultados y colocar el resultado en la posición correspondiente de la matriz resultante.

$$
X = \begin{pmatrix} 1 & 7 \\ 2 & 4 \end{pmatrix}
\qquad
Y = \begin{pmatrix} 3 & 3 \\ 5 & 2 \end{pmatrix}
$$
$$
X \times Y =
\begin{pmatrix}
    1 \times 3 + 7 \times 5 & 1 \times 3 + 7 \times 2 \\
    2 \times 3 + 4 \times 5 & 2 \times 3 + 4 \times 2
\end{pmatrix} =
\begin{pmatrix} 38 & 17 \\ 26 & 14 \end{pmatrix}
$$
Este ejemplo es de [Khan Academy, donde puedes encontrar explicaciones de este proceso desde cero](https://es.khanacademy.org/math/precalculus/x9e81a4f98389efdf:matrices/x9e81a4f98389efdf:multiplying-matrices-by-matrices/a/multiplying-matrices).

Es importante no confundir la multiplicación de matrices con la multiplicación elemento a elemento.

**La multiplicación de matrices es una operación de importancia vital en computación**. Se usa en procesamiento de imágenes, aprendizaje automático, criptografía, compresión de datos, simulación de sistemas físicos, resolución de sistemas de ecuaciones lineales, etc.

```
X = np.array([[1, 7], [2, 4]]); print(f"X =\n{X}")
Y = np.array([[3, 3], [5, 2]]); print(f"Y =\n{Y}")
print("---")
print(f"X @ Y =\n{X @ Y}") # Producto matricial
```

### `@`, `np.matmul` y `np.dot`

`X @ Y` es el operador de Python para el producto matricial, y NumPy lo implementa con `np.matmul`. La función más antigua de NumPy `np.dot` da el mismo resultado con dos matrices, y con dos vectores, donde ambas calculan el producto escalar. Se diferencian en dos casos:

- **Un escalar:** `np.dot(3, X)` multiplica cada elemento por 3, como `3 * X`, mientras que `3 @ X` da un error, porque un escalar no es una matriz.
- **Más de dos dimensiones:** `@` trata cada array como una pila de matrices y las multiplica por parejas, como hacen las bibliotecas de redes neuronales con un lote (*batch*). `np.dot`, en cambio, suma sobre el último eje del primer array y el penúltimo eje del segundo, combinando cada matriz de una pila con cada matriz de la otra.

Usa preferentemente `@` para los productos matriciales: dice lo que significa, y el `@` de PyTorch se comporta igual.

```
np.dot(3, X)
```

```
A = np.ones((2, 3, 4))  # una pila de dos matrices de 3×4
B = np.ones((2, 4, 5))  # una pila de dos matrices de 4×5
print(f"(A @ B).shape = {(A @ B).shape}")
print(f"np.dot(A, B).shape = {np.dot(A, B).shape}")
```
